# Company Sales Data – Matplotlib Exercises
Data: https://pynative.com/wp-content/uploads/2019/01/company_sales_data.csv

```
import pandas as pd
import requests
from io import StringIO

URL = "https://pynative.com/wp-content/uploads/2019/01/company_sales_data.csv"
r = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
r.raise_for_status()
df = pd.read_csv(StringIO(r.text))
df.head(20)
```

## Exercise 1: Total profit of all months (line plot)

```
import matplotlib.pyplot as plt

months = df["month_number"].tolist()

plt.figure(figsize=(8, 5))
plt.plot(months, df["total_profit"], marker="o", color="tab:blue", linewidth=2)
plt.title("Company Profit per Month")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.xticks(months)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 2: Bathing soap and facewash sales (subplots)

```
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 7))

ax1.plot(months, df["bathingsoap"], marker="o", color="tab:green")
ax1.set_title("Bathing Soap Sales Data")
ax1.set_ylabel("Units Sold")
ax1.grid(True, alpha=0.3)

ax2.plot(months, df["facewash"], marker="o", color="tab:red")
ax2.set_title("Facewash Sales Data")
ax2.set_xlabel("Month Number")
ax2.set_ylabel("Units Sold")
ax2.set_xticks(months)
ax2.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```
